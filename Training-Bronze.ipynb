```python
%sql
CREATE TABLE IF NOT EXISTS capgemini_trainingforthecase.bronze.customers
AS SELECT 
    *,
    _metadata.file_path AS _source_file,
    current_timestamp()   AS _ingested_at
FROM read_files(
    '/Volumes/capgemini_challenges/challenge_landing_luan/files/customers.csv',
    format => 'csv'
);
```
