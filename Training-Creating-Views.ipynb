```python
%sql
CREATE OR REPLACE VIEW capgemini_trainingforthecase.gold.vw_top_customers AS
     SELECT customer_name, total_spent, total_orders
     FROM capgemini_trainingforthecase.gold.customer_summary
     ORDER BY total_spent DESC;
```

```python
%sql
CREATE OR REPLACE VIEW capgemini_trainingforthecase.gold.vw_sales_overview AS
SELECT
    order_id,
    order_date,
    customer_name,
    customer_tier,
    product_name,
    category,
    total_amount,
    payment_mode,
    order_status,
    delivery_days
FROM capgemini_trainingforthecase.gold.sales_enriched
ORDER BY order_date DESC;
```

```python
%sql
CREATE OR REPLACE VIEW capgemini_trainingforthecase.gold.vw_top_products AS
SELECT
    product_name,
    category,
    brand,
    total_revenue,
    total_units_sold,
    unique_customers,
    avg_customer_rating
FROM capgemini_trainingforthecase.gold.product_summary
ORDER BY total_revenue DESC;
```
